```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

numerical_columns
```

```python
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"{col}: {len(outliers)} outliers")
```

### Outlier Detection and Treatment

Outliers were detected using the IQR method for the numerical variables.

The boxplots showed that several continuous variables contain extreme
observations and are right-skewed. These observations were not removed because
they may represent genuine differences in online shopping behavior, such as
customers spending significantly more time on product-related pages.

Therefore, no observations were deleted based solely on the IQR outlier
detection method. The original observations were retained to avoid losing
potentially meaningful customer behavior.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.boxplot(
    x='revenue',
    y='pagevalues',
    data=df
)

plt.title('Page Value by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Page Value')

plt.show()
```

### Interpretation

The boxplot shows that Page Value is generally higher for sessions that
resulted in revenue compared with sessions that did not result in revenue.
Non-revenue sessions have Page Value values concentrated close to zero,
whereas revenue sessions show a higher median and greater spread.

This indicates that Page Value may be associated with purchasing behavior.

```python
monthly_purchase_rate = (
    df.groupby(['month', 'visitortype'])['revenue']
      .mean()
      .mul(100)
      .reset_index()
)

monthly_purchase_rate.head()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=monthly_purchase_rate,
    x='month',
    y='revenue',
    hue='visitortype'
)

plt.title('Purchase Rate by Month and Visitor Type')
plt.xlabel('Month')
plt.ylabel('Purchase Rate (%)')
plt.legend(title='Visitor Type')

plt.show()
```
